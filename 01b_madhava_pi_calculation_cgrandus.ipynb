# Actually Solving the Problem 


#### Professor Kwesi Quagraine work to reference (DO NOT RUN)

```python
""" This cell calculates pi via the Madhava-Leibniz series: https://en.wikipedia.org/wiki/Approximations_of_%CF%80#Middle_Ages """

# initialize pi to 0
pi = 0

# set the number of terms in the sum
N = 10000

# initialize an empty list store each term, so we can plot convergence later
pi_terms = []

# loop from 1 to N, adding terms to pi
# such that pi ~= 4*(1 - 1/3 + 1/5 - 1/7 + 1/9 ... -(-1)**n / (2n - 1))
for n in range(1,N):
    if n == 1:
        term = 1
    else:
        term = -((-1)**n) / (2*n - 1)
    pi = pi + term
    
    # store the term
    pi_terms.append(4*pi)

# multiply pi by four to get the final answer
pi = 4 * pi

# print the answer
print(f"pi = {pi:2.7f}")
```

```python
""" Check our answer against the built-in value of pi """
# import the numpy library; give it the shorter name 'np'
import math

# print the true value of pi
print(math.pi)
```

```python
""" Plot the convergence of the series. """

# import the plotting library; give it the shorter name 'plt'
import matplotlib.pyplot as plt

# plot the terms of the series
plt.plot(pi_terms)

# use a log scale on the x-axis
plt.xscale('log')

# plot a horizontal line at the true value of pi
plt.axhline(math.pi, color='gray', linestyle='--')

# set a useful title
plt.title("Convergence of the Madhava-Leibniz series for pi")

# draw the plot
plt.show()
```

## Madhava-Leibniz Approximation of Pi

## 1. Problem - 
#### Question to answer - How does the approximation of π become more accurate as the number of iterations increases?

## 2. Method -
#### Calculate π repeatedly, save each approximation, calculate absolute error, graph the error, then decide how many iterations are accurate enough.

## 3. Investigating Convergence

### 3.1 Calculate Pi Over Many Iterations

```python
pi_sum = 0 # Initilize the variable of pi (where madhava will accumulate)
# pi_sum is only the inside of paranthesis of the equation, you will need to multiply by 4 later
N = 1000 # How many iterations we want to do 
iterations = [] # Empty list to put our variables  
pi_values = [] # Empty list for the pi values after each run

# Loop to run these iterations 
# N + 1 is there because python doesnt include ending number in range()
for n in range(1, N + 1):
    if n == 1: # the equation could do this on its own, but its easier to follow this way
        term = 1
    else:
        term = (-((-1)**n)) / (2*n - 1)

    # add the new series term to the running madhava sum
    pi_sum = pi_sum + term

    # multiply by the 4 so it follows the equation
    current_pi = 4 * pi_sum

    # add the iteration number to the list 
    iterations.append(n)

    # save the current value of pi to the list
    pi_values.append(current_pi)

# print statements/library upload to check
import math

# check how many values were saved (should be the value initilized for N)
print("Number of iterations stored:", len(iterations))
print("Number of pi values stored:", len(pi_values))
print("\n")

# see parts of the lists to verify (not needed just helps me) 
print("First 5 pi values:", pi_values[:5])
print("Last 5 pi values:", pi_values[-5:])
print("\n")

# compare pi
print("Calculated pi:", current_pi)
print("Actual pi:", math.pi)

# I know this is unessesarily long and in future/longer codes its better to minimize...
# but for now it helps me learn/follow aloing

# observation for this section is after 1000 iterations, answer is still about 0.001 away from actual pi
```

### 3.2 Calculate Absolute Error
#### Absolute Error = |Calculated Value - Expected Value|
#### For this assignment Error = |Calculated pi - Actual pi|

```python
# equation to calculate absolute error (use library pi for accuracy)
error = abs(current_pi - math.pi)

# print amount of iterations and the error value
print("Absolute error after", N, "iterations:", error)
```

### 3.3 Plot Error vs. Iterations
#### Make a graph showing how the absolute error changes as the number of iterations increases.

```python
# couldve just added into the original loop in 3.1, but for leanring sake I made a seperate loop/section

errors = [] # new list to compare different error values after each iteration

# calculate absolute error for every saved pi value from the above loop
for value in pi_values:
    error = abs(value - math.pi)
    errors.append(error) # add to the errors list we made above

# print statments for checking 
print("Number of errors stored:", len(errors)) # should be you N value (iterations)
print("\n")
print("First 5 errors:", errors[:5])
print("Last 5 errors:", errors[-5:])
```

```python
# Make the graph 

# was unsure if we were supposed to do the same graph as the solution or make our own, 
# so I made my own in case
import matplotlib.pyplot as plt #import library for graphing

plt.plot(iterations, errors)

# label the graph
plt.xlabel("Number of Iterations")
plt.ylabel("Absolute Error")
plt.title("Convergence of the Madhava-Leibniz Approximation of Pi")

plt.show()
```

```python
# same graph using a log scale for better vizulization

plt.plot(iterations, errors)

plt.xscale("log") #scale we are using

plt.xlabel("Number of Iterations") # lables for the graph 
plt.ylabel("Absolute Error")
plt.title("Convergence of the Madhava-Leibniz Approximation of Pi")

plt.show()
```

#### The figure shows how the absolute error changes as the number of iterations increases. At low iteration counts, the approximation has a relatively large error. As more terms of the Madhava-Leibniz series are included, the absolute error decreases, showing that the numerical approximation is converging toward the accepted value of π.

### 3.4 Reasonable Accuracy

#### I will call my approximation reasonably accurate once it is less than 0.001 away from actual π

```python
tolerance = 0.001 #largest amount of error accepted

for i in range(len(errors)):
    if errors[i] < tolerance:
        required_iterations = iterations[i]# will stop once we get to the acceptable tolerance/threshhold
        break 
        # if we arent allowed to use break in this class please let me know 
        # because some purdue courses allow it and others dont

#print statements
print("Reasonable accuracy: absolute error <", tolerance)
print("Iterations required: ", required_iterations)
```

## 4. Pi Approximation Function

```python
# create a reusable function for this problem 

def calculate_pi(n_iterations): # docstring added below the function
    """
    Calculate an approximation of pi using the Madhava-Leibniz series.
    """
    # initialize 
    pi_sum = 0

    # calculate the madhava series
    for n in range(1, n_iterations + 1): # instead of N we use "n_iterations" beacuse this is a function now
        if n == 1:
            term = 1
        else:
            term = ((-((-1)**n)) / (2*n - 1))

        pi_sum = pi_sum + term  # add the new term to the sum
        current_pi = 4 * pi_sum # make sure to include the 4 from the equation

    # return the final approximation because its a function now
    return current_pi
```

### 4.1 Test the Function

```python
#test the function with any value now that its a function
print("10 iterations:", calculate_pi(10))
print("100 iterations:", calculate_pi(100))
print("1000 iterations:", calculate_pi(1000))
print("\n")
print("Actual pi:", math.pi)
```
